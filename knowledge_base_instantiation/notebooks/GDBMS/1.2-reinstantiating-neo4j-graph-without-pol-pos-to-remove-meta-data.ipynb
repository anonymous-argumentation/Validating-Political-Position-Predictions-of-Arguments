```python
data_base_connection = GraphDatabase.driver(uri = "bolt://localhost:7687", auth=("neo4j", "password"))
session = data_base_connection.session()  
```

```python
with open("../data/processed/GDBMS/nodes_without_political_positions.pickle", "rb") as file:
    node_data = pickle.load(file)
```

```python
for node in node_data:

    if node["utterance_type"] == "___Claim___":
        neo4j_create_statemenet = 'CREATE (:Claim{uuid: '+str(node["uuid"])+', proposition:  "' + node["proposition"] +'", locution: "'+node["locution"]+'", speaker: "'+node["speaker"]+'", utterance_type: "'+node["utterance_type"]+'", illocutionary_force: "'+ node["illocutionary_force"] +'", locution_date: date("'+str(node["locution_date"])+'"), prediction_index: '+str(node["unique_id"])+'});'
    
    elif node["utterance_type"] == "___Question___":
        neo4j_create_statemenet = 'CREATE (:Question{uuid: '+str(node["uuid"])+', proposition:  "' + node["proposition"] +'", locution: "'+node["locution"]+'", speaker: "'+node["speaker"]+'", utterance_type: "'+node["utterance_type"]+'", illocutionary_force: "'+ node["illocutionary_force"] +'", locution_date: date("'+str(node["locution_date"])+'"), prediction_index: '+str(node["unique_id"])+'});'

    elif node["utterance_type"] == "___Concede___":
        neo4j_create_statemenet = 'CREATE (:Concede{uuid: '+str(node["uuid"])+', proposition:  "' + node["proposition"] +'", locution: "'+node["locution"]+'", speaker: "'+node["speaker"]+'", utterance_type: "'+node["utterance_type"]+'", illocutionary_force: "'+ node["illocutionary_force"] +'", locution_date: date("'+str(node["locution_date"])+'"), prediction_index: '+str(node["unique_id"])+'});'

    elif node["utterance_type"] == "___Why___":
        neo4j_create_statemenet = 'CREATE (:Why{uuid: '+str(node["uuid"])+', proposition:  "' + node["proposition"] +'", locution: "'+node["locution"]+'", speaker: "'+node["speaker"]+'", utterance_type: "'+node["utterance_type"]+'", illocutionary_force: "'+ node["illocutionary_force"] +'", locution_date: date("'+str(node["locution_date"])+'"), prediction_index: '+str(node["unique_id"])+'});'

    elif node["utterance_type"] == "Disagreeing":
        neo4j_create_statemenet = 'CREATE (:Disagreeing{uuid: '+str(node["uuid"])+', proposition:  "' + node["proposition"] +'", locution: "'+node["locution"]+'", speaker: "'+node["speaker"]+'", utterance_type: "'+node["utterance_type"]+'", illocutionary_force: "'+ node["illocutionary_force"] +'", locution_date: date("'+str(node["locution_date"])+'"), prediction_index: '+str(node["unique_id"])+'});'

    #print(neo4j_create_statemenet)
    session.run(neo4j_create_statemenet)
```

```python
for rel_from, rel_to, rel_type in relation_data:

    cypher = f'''MATCH (rel1), (rel2) 
    WHERE rel1.uuid = {rel_from["uuid"]} AND rel1.proposition = "{rel_from["proposition"]}" AND rel1.locution = "{rel_from["locution"]}" AND rel1.locution_date = date("{rel_from["locution_date"]}") AND rel1.speaker = "{rel_from["speaker"]}" AND rel1.illocutionary_force = "{rel_from["illocutionary_force"]}"  
    AND rel2.uuid = {rel_to["uuid"]} AND rel2.proposition = "{rel_to["proposition"]}" AND rel2.locution = "{rel_to["locution"]}" AND rel2.locution_date = date("{rel_to["locution_date"]}") AND rel2.speaker = "{rel_to["speaker"]}" AND rel2.illocutionary_force = "{rel_to["illocutionary_force"]}"  
    MERGE (rel1)-[:{rel_type}]->(rel2)'''
    session.run(cypher)
```

```python
from neo4j import GraphDatabase
```

```python
URI = "bolt://localhost:7687"
AUTH = ("neo4j", "password")


with GraphDatabase.driver(URI, auth=AUTH) as driver:
    driver.verify_connectivity()
```

```python
cypher = "MATCH (n) RETURN count(n)"
records_total_num_nodes, summary_total_num_nodes, keys_total_num_nodes = driver.execute_query(cypher)
num_nodes = records_total_num_nodes[0]["count(n)"]
print(num_nodes)
```

```python
for i in range(num_nodes):
    get_data_cypher = 'MATCH (a) WHERE a.unique_id = "'+str(i)+'" RETURN a'
    records, summary, keys = driver.execute_query(get_data_cypher)
    speaker = records[0]["a"]._properties["speaker"] + " : "
    new_locution = records[0]["a"]._properties["locution"]
    new_locution = new_locution.replace(speaker, "")
    speaker = records[0]["a"]._properties["speaker"] + ": "
    #new_locution = records[0]["a"]._properties["locution"]
    new_locution = new_locution.replace(speaker, "")
    change_locution_prop = 'MATCH (a) WHERE a.unique_id = "'+str(i)+'" SET a.locution = "'+new_locution+'" RETURN a'
    driver.execute_query(change_locution_prop)
```
